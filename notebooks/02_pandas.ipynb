Pandas permet de representer des donnees dans des DataFrames

```python
import pandas as pd

donnees = {
    "equipement": ["Routeur A", "Routeur B", "Switch C", "Serveur D"],
    "ville": ["Cotonou", "Bohicon", "Parakou", "Porto-Novo"],
    "temperature": [47, 68, 41, 75],
    "latence": [20, 80, 25, 95]
}


df = pd.DataFrame(donnees)

display(df)

print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.describe())
print(df.info())
```

```python
import pandas as pd

donnees = {
    "equipement": ["Routeur A", "Routeur B", "Switch C", "Serveur D"],
    "ville": ["Cotonou", "Bohicon", "Parakou", "Porto-Novo"],
    "temperature": [47, 68, 41, 75],
    "latence": [20, 80, 25, 95]
}


df = pd.DataFrame(donnees)

print(df[["equipement","temperature"]])
print()
print(df[df["temperature"] > 65])
print()
print(df["latence"].mean())
print()
print(df[(df["temperature"] > 65) & (df["latence"] > 50)])

df["alertes"] = df["temperature"] > 65

display(df)
```
