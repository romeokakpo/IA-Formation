On a les tableaux a double dimensions

```python
import numpy as np

latences = np.array([
    [20, 25, 18, 22],
    [30, 35, 28, 32],
    [15, 18, 20, 17]
])

print("Dimensions:",latences.ndim)
print("Forme:",latences.shape) # (lignes, colonnes)
print("Nombre de valeurs:",latences.size)

# Calcule des statistiques
print("Moyenne globale:", latences.mean())
# print("Moyenne globale:", np.mean(latences))

print("Maximum global:", latences.max())
print("Minimum global:", latences.min())

# Les lignes sont les antiennes et les colonnes sont les periodes

print("Moyenne par ligne:", latences.mean(axis=1)) # np.mean(latences, axis=1)
print("Moyenne par colonne:", latences.mean(axis=0)) # np.mean(latences, axis=0)
```

La notion d'axis est particulièrement importante :
- axis=0 : agrégation selon les lignes, avec un résultat par colonne.
- axis=1 : agrégation selon les colonnes, avec un résultat par ligne.

```python
import numpy as np

v = np.array([1, 2, 3, 4, 5])

print("Moyenne", v.mean())
print(v.shape)
```

Exercice:
Identfication des dimensions latences superieurs a 50

```python
import numpy as np

latences = np.array([20, 25, 18, 80, 22, 95, 19, 30])

anomalies = latences[latences > 50]

print("Anomalies:", anomalies)
print("Pourcentages:", len(anomalies) / len(latences) * 100, "%")
```

Une infrastructure télécom comporte quatre équipements. Les mesures de température, en degrés Celsius, sont :
Équipement	T1	T2	T3
Routeur A	45	48	47
Routeur B	55	62	68
Switch C	39	42	41
Serveur D	60	72	75

Écris un programme NumPy qui :
1. Crée un tableau de dimension (4, 3).
2. Calcule la température moyenne de chaque équipement.
3. Identifie la température maximale globale.
4. Identifie tous les équipements ayant au moins une mesure supérieure à 65 °C.
5. Calcule le pourcentage des 12 mesures supérieures à 65 °C.

```python
import numpy as np

temperatures = np.array([
    [45, 48, 47],
    [55, 62, 68],
    [39, 42, 41],
    [60, 72, 75]
])

noms = np.array(["Routeur A", "Routeur B", "Switch C", "Serveur D"])
seuil = 65

moyennes = np.mean(temperatures,axis = 1)
print("Temperature moyenne de chaque equipement:", moyennes)
print("Temperature max:", np.max(temperatures))

equipements_alertes = np.any(temperatures > seuil, axis=1)

print("Equipement concernes:", noms[equipements_alertes])

temp_65 = temperatures[temperatures > seuil]
print("Pourcentages:", len(temp_65) / temperatures.size * 100, "%")

# Cette methode calcule la moyenne des valeurs positives sur le total
print(np.mean(temperatures > seuil) * 100, '%')
moyennes = np.mean(temperatures,axis = 1)
print("Temperature moyenne la plus elevee:",noms[np.argmax(moyennes)])

print("Nombre de mesures dépassant 65 °C pour chaque équipement", np.sum(temperatures > seuil, axis = 1))

print(noms[np.all(temperatures < seuil, axis= 1)])
```
