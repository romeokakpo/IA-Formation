```python
import numpy as np

# We select all data without the header
taxi = np.genfromtxt("../datasets/nyc_taxis.csv",delimiter=",")[1:]

# Extract pickup month column
pickup_month = taxi[:,1]

# All taxi rides that occurs in january
january = pickup_month[pickup_month == 1]
january_rides = january .shape[0]

# All taxi rides where the tip amount is greater than 20 and total fare is under 50
tip_amount = taxi[:,12]
total_amount = taxi[:,13]

high_tip_low_farebool = (tip_amount > 20) & (total_amount < 50)
result = taxi[high_tip_low_farebool] # On garde uniquement les lignes avec True

print(result)
```
