```python
import pandas as pd

f500 = pd.read_csv("../datasets/f500.csv",index_col=0) # Utilise la premiere colonne comme index
f500.index.name = None

previously_ranked = f500[f500["previous_rank"].notnull()]

revenue_change = previously_ranked["previous_rank"] - previously_ranked["rank"]

f500["rank_change"] = revenue_change

sorted_companies = f500.sort_values("employees", ascending=False)

sorted_companies.head()
```
