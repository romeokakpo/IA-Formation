Data Cleaning

```python
import pandas as pd

laptops = pd.read_csv("../../datasets/laptops.csv", encoding="latin1")

def clean_col(col):
    col = col.strip()
    col = col.replace("Operating System", "os")
    col = col.replace(" ", "_")
    col = col.replace("(", "")
    col = col.replace(")", "")
    col = col.lower()
    return col

# Cleaning columns
new_columns = [clean_col(c) for c in laptops.columns]
laptops.columns = new_columns

"Converting value"
laptops["screen_size"] = (
    laptops["screen_size"].str.replace('"', '').astype(float)
)
# axis = 1 indique que je veux renommer la colonne
laptops.rename({"screen_size": "screen_size_inches"}, axis=1, inplace=True)

# Mettre lrs prix en float
laptops["price_euros"] = laptops["price_euros"].str.replace(",",".").astype(float)

# Remplacer les prix manquants par la moyenne des prix
laptops["price_euros"] = laptops["price_euros"].fillna(laptops["price_euros"].mean())

laptops["gpu"].head().str.split()
```
